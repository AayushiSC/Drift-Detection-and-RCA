```python
import pandas as pd 
import numpy as np
```

```python
df=pd.read_csv('addidas sales data.csv')
df.head()
```

```python
df.columns
```

```python
df['sales'] = df['Total Sales'].replace('[\$,]', '', regex=True).astype(int)
```

```python
df.head()
```

```python
df_walmart=df[df['Retailer']=='Walmart']
```

```python
df_walmart.head()
df_walmart.describe()
```

```python
df_walmart['Invoice Date'] = pd.to_datetime(df_walmart['Invoice Date'])
df_walmart.dtypes
```

```python
df_walmart['year']=df_walmart['Invoice Date'].dt.year
df_walmart['month']=df_walmart['Invoice Date'].dt.month
df_walmart['day']=df_walmart['Invoice Date'].dt.day
df_walmart.head()
```

```python
df_walmart['year'].unique()
```

```python
df_walmart['month'].unique()
```

```python
df_walmart_2020=df_walmart[df_walmart['year']==2020]
df_walmart_2020['month'].unique()

```

```python
df_walmart_2021=df_walmart[df_walmart['year']==2021]
df_walmart_2021['month'].unique()
```

```python
df_walmart_2021=df_walmart_2021[df_walmart_2021['month'].isin([7,8,9,10])]
df_walmart_2021['month'].unique()
```

```python
#Retailer name is walmart 
# set 1 is july-aug-sept-oct 2020
# set 2 is july-aug-sept-oct 2021
df_walmart_2021.to_csv('new_data.csv',index=False)
df_walmart_2020.to_csv('old_data.csv',index=False)
```
