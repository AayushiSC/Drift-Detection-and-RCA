```python
import pandas as pd
```

```python
df=pd.read_csv('yellow_tripdata_2016-01.csv')
df.head()
```

```python
df.shape
```

```python
df.dtypes
```

```python
df['tpep_pickup_datetime']=pd.to_datetime(df['tpep_pickup_datetime'])
df['date']=df['tpep_pickup_datetime'].dt.date
df['date'].nunique()
```

```python
df.columns
```

```python
# Group by date and sum numeric columns (excluding latitude/longitude)
# First, let's identify columns to exclude
exclude_cols = ['tpep_pickup_datetime', 'tpep_dropoff_datetime', 'date', 
                'pickup_longitude', 'pickup_latitude', 
                'dropoff_longitude', 'dropoff_latitude']

# Get numeric columns to aggregate
numeric_cols = df.select_dtypes(include=['int64', 'float64']).columns.tolist()
agg_cols = [col for col in numeric_cols if col not in exclude_cols]

# Group by date and sum
df_daily = df.groupby('date')[agg_cols].sum().reset_index()
df_daily.head()
```

```python
df_daily.shape
```

```python
df_daily.to_csv('new_data.csv', index=False)
```
